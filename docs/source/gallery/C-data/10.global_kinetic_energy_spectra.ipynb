# Kinetic energy spectra for global unstretched grids

For global unstretched grids, kinetic energy spectra can be derived directly from the spectral coefficients of streamfunction and velocity potential. More precisely, the kinetic energy per unit mass can be decomposed into a rotational and a divergent component, associated to the energy spectra of vorticity and divergence, respectively. 

Reference : Koshyk, J. N., and K. Hamilton, 2001: *The Horizontal Kinetic Energy Spectrum and Spectral Budget Simulated by a High-Resolution Troposphere–Stratosphere–Mesosphere GCM*. J. Atmos. Sci., 58, 329–348 [doi:10.1175/1520-0469(2001)058<0329:THKESA>2.0.CO;2](https://doi.org/10.1175/1520-0469(2001)058<0329:THKESA>2.0.CO;2).

```python
%matplotlib inline
# for figures in notebook

# Import & initialize epygram
import epygram
import epygram.spectra as sp
epygram.init_env()

# Open FA resource
import pathlib
res = epygram.open(pathlib.Path("..") / "inputs" / "historic.arpege.tl149-c10+0003:00.fa", "r")
```

The `get_global_energy_spectra` returns a tuple of three `epygram.Spectrum` objects, associated respectively to the total kinetic energy, its divergent component and its rotational component. 

```python
# Perform spectral analysis
ke, dke, rke = sp.get_global_energy_spectra(res, level=80)

# Plot spectra. The spectra already have meaningful names, used as labels by `plotspectra`.
fig, ax = sp.plotspectra(
    [ke, rke, dke],
    figsize=(6, 5),
    unit="\\text{energy per unit mass}",
    title="3h forecast on TL149 grid",
)
```
